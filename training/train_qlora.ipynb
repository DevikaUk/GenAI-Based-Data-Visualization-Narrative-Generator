```python
def validate_narrative(narrative, insights):
    """
    Basic validation to check whether important
    numerical values from the structured insights
    appear in the generated narrative.
    """

    important_values = []

    for metric in insights.get("metrics", []):
        important_values.append(str(metric["value"]))

    for trend in insights.get("trends", []):
        important_values.append(str(trend["change"]))

    for comparison in insights.get("comparisons", []):
        important_values.append(str(comparison["highest"]["value"]))
        important_values.append(str(comparison["lowest"]["value"]))

    missing = []

    for value in important_values:
        if value not in narrative:
            missing.append(value)

    return {
        "valid": len(missing) == 0,
        "missing_values": missing
    }
```

```python
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

tokenizer = AutoTokenizer.from_pretrained(model_name)

base_model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto"
)

finetuned_model = PeftModel.from_pretrained(
    base_model,
    "qwen_lora_output"
)

print("Fine-tuned QLoRA model loaded.")
```

```python
def generate_narrative(model, insights, max_new_tokens=300):

    insights_json = json.dumps(insights, indent=2)

    prompt = f"""
You are a business analytics assistant.

Generate a concise analytical narrative from the structured
analytical findings provided below.

Rules:
1. Use ONLY the information provided.
2. Do not invent numbers, facts, trends, or causes.
3. Mention the most important metrics.
4. Mention important trends.
5. Mention important comparisons.
6. Mention anomalies if they are provided.
7. Ignore empty sections.
8. Do not claim that one factor caused another unless causation
   is explicitly provided.
9. Write in a professional business-analysis style.

Structured analytical findings:
{insights_json}

Generate the business narrative now.
"""

    messages = [
        {"role": "user", "content": prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(text, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False
        )

    generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )
```

```python
def calculate_coverage(narrative, supported_facts):
    narrative_lower = narrative.lower()

    found = []
    missing = []

    for fact in supported_facts:
        if fact.lower() in narrative_lower:
            found.append(fact)
        else:
            missing.append(fact)

    coverage = len(found) / len(supported_facts) * 100

    return coverage, found, missing
```

```python
import re
import json

def extract_numbers(text):
    return set(re.findall(r'\b\d+(?:\.\d+)?\b', text))

def factual_consistency(narrative, insights):
    input_numbers = extract_numbers(json.dumps(insights))
    output_numbers = extract_numbers(narrative)

    unsupported_numbers = output_numbers - input_numbers

    if len(output_numbers) == 0:
        return 100.0

    supported_numbers = output_numbers - unsupported_numbers

    return (len(supported_numbers) / len(output_numbers)) * 100
```

```python
def relevance_score(narrative, supported_facts):
    narrative_lower = narrative.lower()

    relevant_facts = [
        fact for fact in supported_facts
        if fact.lower() in narrative_lower
    ]

    if len(supported_facts) == 0:
        return 100.0

    return (len(relevant_facts) / len(supported_facts)) * 100
```

```python
baseline_relevance = relevance_score(
    baseline_response,
    supported_facts
)

qlora_relevance = relevance_score(
    finetuned_response,
    supported_facts
)

evaluation_df.loc[0, "Relevance"] = baseline_relevance
evaluation_df.loc[1, "Relevance"] = qlora_relevance

evaluation_df
```

```python
def validate_narrative(narrative, insights):

    # Numbers present in the source insights
    input_numbers = extract_numbers(json.dumps(insights))

    # Numbers present in generated narrative
    output_numbers = extract_numbers(narrative)

    # Numbers that were not present in the source
    unsupported_numbers = output_numbers - input_numbers

    # Check if all source numbers are covered
    missing_numbers = input_numbers - output_numbers

    # Narrative is valid only if there are no unsupported numbers
    is_valid = len(unsupported_numbers) == 0

    return {
        "is_valid": is_valid,
        "unsupported_numbers": sorted(unsupported_numbers),
        "missing_numbers": sorted(missing_numbers)
    }
```
